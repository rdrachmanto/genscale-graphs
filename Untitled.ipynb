```python
import pandas as pd
import matplotlib.pyplot as plt

# -----------------------------------
# Pretend runtime data
# -----------------------------------
data = [
    # BWA-MEM: relatively stable across CPUs
    ["BWA-MEM", 0.1, "AMD EPYC 7352", 10],
    ["BWA-MEM", 0.1, "Intel Xeon 6126", 11],
    ["BWA-MEM", 0.1, "Intel Xeon 6240R", 10.5],

    ["BWA-MEM", 0.5, "AMD EPYC 7352", 40],
    ["BWA-MEM", 0.5, "Intel Xeon 6126", 42],
    ["BWA-MEM", 0.5, "Intel Xeon 6240R", 41],

    ["BWA-MEM", 1.0, "AMD EPYC 7352", 75],
    ["BWA-MEM", 1.0, "Intel Xeon 6126", 78],
    ["BWA-MEM", 1.0, "Intel Xeon 6240R", 76],

    ["BWA-MEM", 2.0, "AMD EPYC 7352", 140],
    ["BWA-MEM", 2.0, "Intel Xeon 6126", 145],
    ["BWA-MEM", 2.0, "Intel Xeon 6240R", 142],

    ["BWA-MEM", 4.0, "AMD EPYC 7352", 270],
    ["BWA-MEM", 4.0, "Intel Xeon 6126", 278],
    ["BWA-MEM", 4.0, "Intel Xeon 6240R", 274],

    ["BWA-MEM", 8.0, "AMD EPYC 7352", 520],
    ["BWA-MEM", 8.0, "Intel Xeon 6126", 535],
    ["BWA-MEM", 8.0, "Intel Xeon 6240R", 528],

    # PCWM: more variation across CPUs, especially for larger inputs
    ["PCWM", 0.1, "AMD EPYC 7352", 15],
    ["PCWM", 0.1, "Intel Xeon 6126", 16],
    ["PCWM", 0.1, "Intel Xeon 6240R", 17],

    ["PCWM", 0.5, "AMD EPYC 7352", 60],
    ["PCWM", 0.5, "Intel Xeon 6126", 68],
    ["PCWM", 0.5, "Intel Xeon 6240R", 72],

    ["PCWM", 1.0, "AMD EPYC 7352", 120],
    ["PCWM", 1.0, "Intel Xeon 6126", 145],
    ["PCWM", 1.0, "Intel Xeon 6240R", 155],

    ["PCWM", 2.0, "AMD EPYC 7352", 230],
    ["PCWM", 2.0, "Intel Xeon 6126", 290],
    ["PCWM", 2.0, "Intel Xeon 6240R", 310],

    ["PCWM", 4.0, "AMD EPYC 7352", 440],
    ["PCWM", 4.0, "Intel Xeon 6126", 570],
    ["PCWM", 4.0, "Intel Xeon 6240R", 610],

    ["PCWM", 8.0, "AMD EPYC 7352", 850],
    ["PCWM", 8.0, "Intel Xeon 6126", 1120],
    ["PCWM", 8.0, "Intel Xeon 6240R", 1180],
]

df = pd.DataFrame(data, columns=["tool", "input_size_gb", "cpu", "runtime_sec"])

# -----------------------------------
# Compute cross-platform variation
# -----------------------------------
variation_df = (
    df.groupby(["tool", "input_size_gb"])["runtime_sec"]
    .agg(["min", "max", "mean"])
    .reset_index()
)

variation_df["variation_percent"] = (
    (variation_df["max"] - variation_df["min"]) / variation_df["mean"] * 100
)

# print(variation_df)

# -----------------------------------
# Plot
# -----------------------------------
fig, ax = plt.subplots(figsize=(6, 4))

tool_order = ["PCWM", "BWA-MEM"]
colors = {
    "PCWM": "#CC79A7",
    "BWA-MEM": "#0072B2"
}
linestyles = {
    "PCWM": "-",
    "BWA-MEM": "--"
}

for tool in tool_order:
    sub = variation_df[variation_df["tool"] == tool]
    ax.plot(
        sub["input_size_gb"],
        sub["variation_percent"],
        marker="o",
        linewidth=2,
        markersize=5,
        linestyle=linestyles[tool],
        color=colors[tool],
        label=tool
    )

ax.set_xlabel("Input Size (GB)", fontsize=11)
ax.set_ylabel("Cross-platform Variation (%)", fontsize=11)
ax.grid(axis="y", alpha=0.3)
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.tick_params(axis="both", labelsize=10, length=0)
ax.legend(frameon=False)

plt.tight_layout()

plt.savefig("proto")

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

# Dummy relative-error data
# Most values are small, with some large outliers
data = {
    "PTS-Hybrid": np.concatenate([
        np.random.gamma(shape=2.0, scale=0.08, size=180),
        np.array([1.5, 2.8, 4.2, 5.1])
    ]),
    "Lotaru-G": np.concatenate([
        np.random.gamma(shape=2.2, scale=0.10, size=180),
        np.array([1.8, 2.4, 3.7, 5.3])
    ]),
    "Lotaru-A": np.concatenate([
        np.random.gamma(shape=2.5, scale=0.12, size=180),
        np.array([2.0, 3.2, 4.6, 7.3])
    ]),
    "OnlineP": np.concatenate([
        np.random.gamma(shape=2.8, scale=0.13, size=180),
        np.array([3.1, 6.7, 9.8, 12.0, 14.1])
    ])
}
```

```python
labels = list(data.keys())
values = [data[k] for k in labels]

fig, (ax_top, ax_bottom) = plt.subplots(
    2, 1,
    sharex=True,
    figsize=(7, 5),
    gridspec_kw={"height_ratios": [1, 3]}
)

# Draw boxplots on both axes
for ax in (ax_top, ax_bottom):
    ax.boxplot(
        values,
        labels=labels,
        patch_artist=True,
        showfliers=True,
        boxprops=dict(linewidth=1.2),
        medianprops=dict(linewidth=1.5),
        whiskerprops=dict(linewidth=1.2),
        capprops=dict(linewidth=1.2),
        flierprops=dict(
            marker="o",
            markersize=4,
            markerfacecolor="none",
            markeredgecolor="black",
            linestyle="none"
        )
    )

# Bottom shows the box/median region
ax_bottom.set_ylim(0, 0.7)

# Top shows the outlier region
ax_top.set_ylim(1.2, 15)

# Hide spines between axes
ax_top.spines["bottom"].set_visible(False)
ax_bottom.spines["top"].set_visible(False)

ax_top.tick_params(labeltop=False)
ax_bottom.xaxis.tick_bottom()

# Add diagonal break marks
d = 0.015

kwargs = dict(transform=ax_top.transAxes, color="black", clip_on=False)
ax_top.plot((-d, +d), (-d, +d), **kwargs)
ax_top.plot((1 - d, 1 + d), (-d, +d), **kwargs)

kwargs.update(transform=ax_bottom.transAxes)
ax_bottom.plot((-d, +d), (1 - d, 1 + d), **kwargs)
ax_bottom.plot((1 - d, 1 + d), (1 - d, 1 + d), **kwargs)

ax_top.set_ylabel("Outliers")
ax_bottom.set_ylabel("Relative Error")
ax_bottom.set_xlabel("Estimator")

fig.suptitle("Relative Error Distribution", y=0.98)
plt.tight_layout()
plt.show()
```
